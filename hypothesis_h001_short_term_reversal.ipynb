## ١١) فحص نافذة بعينها — تشخيص أنظمة السوق (النافذتان الشاذّتان ١٤، ١٨)

أداتان عامّتان (تُستخدَمان لأي نافذة لاحقاً، لا لهاتين فقط): `inspect_window`
تستخرج تواريخ ونِسَب `train`/`val`/`test` الفعلية من `windows[n-1]` مباشرة —
بلا حاجة لحفظ `schedule` بمعزل. `regime_snapshot` تُعطي لقطة خام (اتجاه عام
+ تقلّب "آخر تغيّر") لعيّنات `test` في تلك النافذة — تشخيص أوّلي فقط، لا
بديل عن نظام أنظمة سوق كامل (HMM/GMM، المرحلة ٢ في خطة المشروع).

تتطلّبان `sample_timestamps`/`embargo_duration` (من دفتر التحضير) و
`concat_splits`/`extract_feature_last_diff` (قسم ٦ أعلاه) متاحة في نفس
الجلسة — كبقية أمثلة هذا القسم.

```python
# @title
def inspect_window(windows, n: int, window_names=None) -> dict:
    """تواريخ ونِسَب train/val/test الفعلية للنافذة رقم n (١-محسوبة) من
    windows الناتجة عن rolling_splits — نفس القائمة المُمرَّرة أصلاً لـ
    evaluate_hypothesis_over_rolling_windows، بلا حاجة لحفظ schedule بمعزل."""
    if not (1 <= n <= len(windows)):
        raise IndexError(f"النافذة {n} خارج المدى (المتاح 1..{len(windows)}).")
    train, val, test = windows[n - 1]
    name = window_names[n - 1] if window_names else f"نافذة {n}"

    tr_ts = sample_timestamps(concat_splits(train))
    va_ts = sample_timestamps(concat_splits(val))
    te_ts = sample_timestamps(concat_splits(test))

    tr_start, tr_end, tr_n = tr_ts.min(), tr_ts.max(), len(tr_ts)
    va_start, va_end, va_n = va_ts.min(), va_ts.max(), len(va_ts)
    te_start, te_end, te_n = te_ts.min(), te_ts.max(), len(te_ts)

    info = {"name": name,
            "train_range": (tr_start, tr_end, tr_n),
            "val_range": (va_start, va_end, va_n),
            "test_range": (te_start, te_end, te_n)}
    print(f"🔁 {name}: "
          f"train {tr_start:%Y-%m-%d}→{tr_end:%Y-%m-%d} ({tr_n:,}) | "
          f"val {va_start:%Y-%m-%d}→{va_end:%Y-%m-%d} ({va_n:,}) | "
          f"test {te_start:%Y-%m-%d}→{te_end:%Y-%m-%d} ({te_n:,})")
    return info


def regime_snapshot(windows, n: int, feature: str = "close", tf=None,
                    feature_order=None) -> dict:
    """لقطة نظام سوقي خام لعيّنات test في النافذة n: اتجاه عام (متوسط 'آخر
    تغيّر' في السعر) وتقلّبه (انحراف معياري) — يفحص هل نافذة شاذّة (IC معكوس
    عن بقية النوافذ) تقع في سوق متّجه بقوة أو متذبذب. مرّر feature_order من
    dataset['feature_order'] الأصلي (غير محفوظة داخل النافذة نفسها)."""
    _, _, test = windows[n - 1]
    diffs = extract_feature_last_diff(test, feature=feature, tf=tf, feature_order=feature_order)
    mean_diff = float(np.mean(diffs))
    std_diff = float(np.std(diffs))
    frac_pos = float(np.mean(diffs > 0))
    snap = {"n_samples": len(diffs), "mean_last_diff": mean_diff,
           "std_last_diff": std_diff, "frac_positive": frac_pos}
    print(f"   اتجاه عام={mean_diff:+.4f} | تقلّب={std_diff:.4f} | "
          f"نسبة عيّنات صاعدة={frac_pos:.0%}")
    return snap


# شغّل هذا بعد بناء windows (قسم ٨) في نفس الجلسة:
for n in (14, 18):
    inspect_window(windows, n)
    regime_snapshot(windows, n, feature_order=dataset["feature_order"])
```

## ١٣) تصحيح: فرضية "قرب حدث سوقي حاسم" لا تصمد أمام خط الأساس

مقارنة فعلية بالأرقام (نوافذ ١٠/٢١/٢٦ هي **أقوى ثلاث نوافذ انعكاس** — IC
الأكثر سلبية بين الثلاثين):

| النافذة | IC | اتجاه عام | تقلّب | % صاعدة |
|---|---|---|---|---|
| ١٤ (شاذّة) | +0.080 | +0.0070 | **0.5064** | 49% |
| ١٨ (شاذّة) | +0.087 | -0.0372 | 0.3195 | 52% |
| ١٠ (أساس) | -0.180 | -0.0617 | 0.4948 | 45% |
| ٢١ (أساس) | -0.189 | -0.0914 | 0.4341 | 41% |
| ٢٦ (أساس) | -0.215 | -0.0337 | 0.4617 | 45% |

لا مقياس من الثلاثة يفصل ١٤/١٨ عن خط الأساس باتجاه ثابت:

* **التقلّب**: نوافذ الأساس كلها بين 0.43-0.49. نافذة ١٤ (0.51) بالكاد
  أعلاها — فرق هامشي لا يستحقّ وصفه "الأعلى بوضوح" كما قلتُ سابقاً. نافذة
  ١٨ (0.32) **أقلّ** من كل نوافذ الأساس — عكس اتجاه فرضية "تقلّب أعلى" تماماً.
* **الاتجاه العام**: نافذة ١٨ (-0.037) شبه مطابقة لنافذة الأساس ٢٦ (-0.034) —
  نفس الاتجاه والحجم تقريباً، لكن IC متعاكس تماماً (+0.087 مقابل -0.215).
  لو كان الاتجاه العام هو المُفسِّر، لاتّفقت هاتان النافذتان في IC، ولم
  يحدث.
* **% صاعدة**: نافذة ١٨ (52%) فعلاً أعلى من كل نوافذ الأساس (41-45%) — أقرب
  إشارة تمييز حقيقية وُجدت، لكنها وحيدة (لا تدعمها نافذة ١٤: 49%، قريبة من
  المدى العادي) ومن بيانات قليلة جداً (نافذتان فقط) لبناء استنتاج عليها.

**الخلاصة الصادقة: التفسير الذي اقترحتُه ("قرب حدث سوقي حاسم") مفنَّد فعلياً
بهذه المقارنة — لا يفسّر البيانات.** الشذوذ (نافذتان من 30 بإشارة معاكسة
قوية) لا يزال **غير مُفسَّر** بإحصاءات `regime_snapshot` الخام (اتجاه/تقلّب/
نسبة صاعدة). هذا لا يُضعف H001 نفسها — 28/30 نافذة (93%) لا تزال متّسقة،
والدليل الإحصائي على الانعكاس قوي بذاته — لكنه يعني أن "لماذا هاتان
النافذتان تحديداً" سؤال مفتوح حقيقي، لا محلولاً كما بدا سابقاً.

**الخطوة الصحيحة التالية إن أردتَ ملاحقته فعلياً**: ليست مزيداً من لقطات
يدوية لنوافذ مُختارة يدوياً (تعرّض حقيقي لتحيّز اختيار العيّنة) — بل أداة
كشف الأنظمة غير المُشرَفة (HMM/GMM على التقلّب والاتجاه) المُدرَجة أصلاً في
خطة المشروع، قسم "أدوات ونماذج لاكتشاف نمط غير معروف"، مرحلة ٣. أرخص بديل
عملي إن أردتَ الاستمرار الآن دون بناء ذلك: تشغيل `regime_snapshot` على
الثلاثين نافذة كلها (لا خمس مُنتقاة) ثم فحص هل ١٤/١٨ فعلاً قيمتان متطرّفتان
إحصائياً على أي عمود (percentile خارج التوزيع الكامل)، لا مجرّد مقارنة بثلاث
نوافذ يدوية.

```python
# @title
# خيار أسرع من بناء HMM/GMM كاملة: شغّل regime_snapshot على الثلاثين نافذة
# كلها (لا خمس مُنتقاة يدوياً)، ثم افحص أين تقع 14/18 فعلياً ضمن التوزيع
# الكامل — تحيّز اختيار العيّنة هو أكبر ثغرة في المقارنة اليدوية أعلاه.
_all_snaps = []
for n in range(1, len(windows) + 1):
    s = regime_snapshot(windows, n, feature_order=dataset["feature_order"])
    s["n"] = n
    _all_snaps.append(s)

import pandas as pd
_regime_table = pd.DataFrame(_all_snaps)[["n", "n_samples", "mean_last_diff", "std_last_diff", "frac_positive"]]
_regime_table["std_percentile"] = _regime_table["std_last_diff"].rank(pct=True) * 100
_regime_table["mean_percentile"] = _regime_table["mean_last_diff"].rank(pct=True) * 100
print(_regime_table.to_string(index=False))
print()
print("نافذة 14 — مئينيّ التقلّب:", _regime_table.loc[_regime_table["n"] == 14, "std_percentile"].iloc[0])
print("نافذة 18 — مئينيّ التقلّب:", _regime_table.loc[_regime_table["n"] == 18, "std_percentile"].iloc[0])
```

### تصحيح عدّ من الملاحظة الأولى في هذه المحادثة

"22 من 30 نافذة سالبة" (أول رسالة) كان خطأً — خُلط مع عدد النوافذ
**السالبة والمعنوية معاً**. العدّ الدقيق من `per_window` الكامل:

| | العدد |
|---|---|
| إشارة سالبة (أي درجة) | 25/30 |
| إشارة موجبة (أي درجة) | 5/30 — نوافذ ٢، ١٤، ١٧، ١٨، ٢٧ |
| سالبة **ومعنوية** (p<0.05) | 22/30 |
| موجبة **ومعنوية** (p<0.05) | **٢/30 فقط — نافذتا ١٤، ١٨** |

النوافذ الموجبة الثلاث الأخرى (٢، ١٧، ٢٧) غير معنوية (p=0.43، 0.38، 0.16) —
ضجيج محض، لا شذوذ حقيقي مثل ١٤/١٨.

```python
# @title
from scipy import stats

_ic_by_n = {int(r["window"].split()[-1]): r["ic"] for r in _per_window_actual}
_ic_series = [_ic_by_n[n] for n in _regime_table["n"]]

for col in ("mean_last_diff", "std_last_diff", "frac_positive"):
    r_s, p_s = stats.spearmanr(_ic_series, _regime_table[col])
    r_p, p_p = stats.pearsonr(_ic_series, _regime_table[col])
    print(f"{col:16s} spearman r={r_s:+.3f} (p={p_s:.3f})   pearson r={r_p:+.3f} (p={p_p:.3f})")

import numpy as np
_ic_arr = np.array(_ic_series)
_pos = _ic_arr > 0
print()
print("متوسط التقلّب | IC>0:", _regime_table.loc[_pos, "std_last_diff"].mean(),
     " | IC<0:", _regime_table.loc[~_pos, "std_last_diff"].mean())
```
